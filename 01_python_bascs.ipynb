```python
# 첫 번째 KBO 예측 프로젝트 테스트 코드
team_name = "한화 이글스"
print(f"안녕하세요! {team_name} 데이터 분석을 시작합니다.")

# 간단한 덧셈 테스트
wins = 57
losses = 83
draw = 4
total_games = wins + losses + draw
print(f"총 경기 수: {total_games}경기")
```

```python
# KBO 팀 승률에 따른 등급 판정 프로그램 예제

team_name = "한화 이글스"
wins = 57
losses = 83
draws = 4

# 1. 총 경기 수 및 승률 계산 (무승부는 보통 승패 계산에서 제외하거나 다르게 처리하지만 간단히 계산해봅니다)
total_games = wins + losses
win_rate = wins / total_games

print(f"[{team_name}] 성적 집계")
print(f"승: {wins} / 패: {losses} / 승률: {win_rate:.3f}") # :.3f는 소수점 셋째 자리까지 출력하라는 뜻입니다.

# 2. 조건문(if-elif-else)을 이용한 등급 나누기
if win_rate >= 0.550:
    print(result := f"판정: 🔥 {team_name}은(는) 상위권 팀입니다!")
elif win_rate >= 0.480:
    print(result := f"판정: ⚡ {team_name}은(는) 중위권 치열한 구간입니다.")
else:
    print(result := f"판정: 🌧️ {team_name}은(는) 하위권입니다. 반등이 필요해요!")
```

```python
# =========================================================
# KBO 10개 구단 전체 성적 데이터 다루기 (반복문 for 활용)
# =========================================================

# 1. KBO 10개 구단 성적 데이터 (리스트와 딕셔너리 구조)
kbo_teams = [
    {"team": "KIA", "wins": 79, "losses": 63, "draw": 2},
    {"team": "삼성", "wins": 83, "losses": 58, "draw": 3},
    {"team": "LG", "wins": 78, "losses": 65, "draw": 1},
    {"team": "두산", "wins": 73, "losses": 66, "draw": 5},
    {"team": "KT", "wins": 90, "losses": 49, "draw": 5},
    {"team": "SSG", "wins": 65, "losses": 74, "draw": 5},
    {"team": "롯데", "wins": 64, "losses": 77, "draw": 3},
    {"team": "한화", "wins": 57, "losses": 83, "draw": 4},
    {"team": "NC", "wins": 64, "losses": 78, "draw": 2},
    {"team": "키움", "wins": 50, "losses": 90, "draw": 4}
]

print("⚾ 2026 KBO 전 구단 성적 및 등급 자동 분석 결과")
print("=" * 55)

# 2. 반복문(for)을 사용해 10개 팀을 하나씩 꺼내어 자동 처리하기
for t in kbo_teams:
    name = t["team"]
    w = t["wins"]
    l = t["losses"]

    # 승률 계산
    win_rate = w / (w + l)

    # 등급 판정
    if win_rate >= 0.550:
        grade = "🔥 A등급 (상위권)"
    elif win_rate >= 0.480:
        grade = "⚡ B등급 (중위권)"
    else:
        grade = "🌧️ C등급 (하위권)"

    # 결과 출력
    print(f"[{name}] {w}승 {l}패 | 승률: {win_rate:.3f} | {grade}")

print("=" * 55)
print("✅ 전 구단 분석 완료!")
```
