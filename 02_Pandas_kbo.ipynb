```python
# 파이썬의 강력한 데이터 분석 도구 '판다스(Pandas)' 불러오기
import pandas as pd

# 판다스가 잘 설치되어 있는지 확인하는 간단한 가상 엑셀 표 만들기
data = {
    '팀명': ['KT', '삼성', 'KIA', 'LG'],
    '승': [90, 83, 79, 78],
    '패': [49, 58, 63, 65]
}

# 파이썬의 딕셔너리를 판다스 '데이터프레임(DataFrame, 엑셀 표 형태)'으로 변환하기
df = pd.DataFrame(data)

print("📊 판다스로 만든 가상의 KBO 성적 표:")
print(df)
print("-" * 30)

# 승률 계산 열(Column)을 표에 통째로 추가하기
df['승률'] = df['승'] / (df['승'] + df['패'])

print("🔥 승률 계산 열이 추가된 최종 표:")
print(df)
```
